```python
import kagglehub
import os

path = kagglehub.dataset_download(
    "ziya07/indian-languages-audio-dataset"
)

print("Dataset path:")
print(path)

print("\nFiles:")
for root, dirs, files in os.walk(path):
    level = root.replace(path, "").count(os.sep)
    indent = "  " * level
    print(f"{indent}{os.path.basename(root)}/")

    for file in files[:10]:
        print(f"{indent}  {file}")
```

```python
from pathlib import Path

DATASET_PATH = Path(path)

audio_extensions = {
    ".wav",
    ".mp3",
    ".flac",
    ".ogg",
    ".m4a",
    ".aac"
}

audio_files = []

for file in DATASET_PATH.rglob("*"):
    if file.is_file() and file.suffix.lower() in audio_extensions:
        audio_files.append(file)

print("Total audio files:", len(audio_files))

for file in audio_files[:20]:
    print(file)
```

```python
from collections import Counter

extensions = Counter(
    file.suffix.lower()
    for file in audio_files
)

print("Audio format distribution:")

for ext, count in extensions.items():
    print(f"{ext}: {count}")
```

```python
import librosa
import pandas as pd
from tqdm.auto import tqdm

records = []

for file in tqdm(audio_files):

    try:
        audio, sr = librosa.load(file, sr=None, mono=True)

        duration = librosa.get_duration(
            y=audio,
            sr=sr
        )

        records.append({
            "audio_path": str(file),
            "filename": file.name,
            "extension": file.suffix.lower(),
            "sample_rate": sr,
            "duration": duration,
            "samples": len(audio)
        })

    except Exception as e:

        records.append({
            "audio_path": str(file),
            "filename": file.name,
            "extension": file.suffix.lower(),
            "sample_rate": None,
            "duration": None,
            "samples": None,
            "error": str(e)
        })

df = pd.DataFrame(records)

print(df.head())
print("\nDataset shape:", df.shape)
```

```python
print("Total files:", len(df))

print(
    "\nTotal duration:",
    df["duration"].sum() / 3600,
    "hours"
)

print(
    "\nAverage duration:",
    df["duration"].mean(),
    "seconds"
)

print(
    "\nMinimum duration:",
    df["duration"].min(),
    "seconds"
)

print(
    "\nMaximum duration:",
    df["duration"].max(),
    "seconds"
)
```
